```python
import numpy as np
import pandas as pd
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def activation(number):
    return 1 if number > 0 else 0
```

```python
class MLP:
    # create wt graph with random weights
    def __init__(self, input_size, hidden_size, output_size):
        # initalize mlp with random weights
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
    
        # Weights and biases
        self.weights_input_hidden = np.random.uniform(-1, 1, (input_size, hidden_size))
        self.bias_hidden = np.ones((1, hidden_size))
        self.weights_hidden_output = np.random.uniform(-1, 1, (hidden_size, output_size))
        self.bias_output = np.ones((1, output_size))

    def feedforward(self, inputs):
        # Input to hidden layer
        self.hidden_layer_input = np.dot(inputs, self.weights_input_hidden) + self.bias_hidden
        self.hidden_layer_output = sigmoid(self.hidden_layer_input)

        # Hidden to output layer
        self.output_layer_input = np.dot(self.hidden_layer_output, self.weights_hidden_output) + self.bias_output
        self.output_layer_output = sigmoid(self.output_layer_input)

        return self.output_layer_output

    def train(self, inputs, expected_output, epochs):
        for epoch in range(epochs):
            # Forward pass
            self.feedforward(inputs)

            if epoch % 1000 == 0:
                loss = np.mean((expected_output - self.output_layer_output) ** 2)
                print(f"Epoch {epoch}, Loss: {loss:.4f}")

    def predict(self, inputs):
        return self.feedforward(inputs)
        
```

```python
inputs = np.array([[0,0], [0,1], [1,0], [1,1]])
output = np.array([[0], [1], [1], [0]])
```

```python
mlp = MLP(2, 2, 1)
```

```python
mlp.train(inputs, output, 100000)
```

```python
mlp.predict([1, 0])
```

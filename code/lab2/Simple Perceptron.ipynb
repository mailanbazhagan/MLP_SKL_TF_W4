## Function

```python
def compute_weighted_sum(X, weights, b):
    Z = []
    for idx in range(len(X)):
        Z.append(np.dot(X[idx], weights) + b)

    return Z
```

```python
def activation_function(Z):
    Y = []
    for idx in range(len(Z)):
        if(Z[idx] >= 0):
            Y.append(1)
        else:
            Y.append(0)

    return Y
```

```python
def update_weights(Y_pred, Y, X, weights, idx):
    n = 0.001
    
    weights[0] = weights[0] + (n * (Y[idx] - Y_pred[idx])) * X[idx][0]
    weights[1] = weights[1] + (n * (Y[idx] - Y_pred[idx])) * X[idx][1]

    return weights
```

```python
def update_bias(Y_pred, Y, b, idx):
    n = 0.1
    
    return b + n*(Y[idx] - Y_pred[idx]) 
```

```python
def perceptron(dataset_x, dataset_y, weights, b):
    Z = []
    Y_pred = []
    flag = True
    past_error = 0
    idx = 0
    
    while(flag):
        Z = compute_weighted_sum(dataset_x, weights, b)
        Y_pred = activation_function(Z)

        if(Y_pred[idx] != dataset_y[idx] or (past_error != (Y_pred[idx] - dataset_y[idx]))): 
            past_error = Y_pred[idx] - dataset_y[idx]
            weights = update_weights(Y_pred, dataset_y, dataset_x, weights, idx)
            b = update_bias(Y_pred, dataset_y, b, idx)
        else:
            idx += 1

        if(idx == 4):
            flag = False
            

    return (Y_pred, weights, b)
    
```
